```python
text = "My World. Nice World. A happy World! Indeed."
```

```python
def all_positions(haystack, needle):
    pos, hits = -1, []
    while True:
        pos = haystack.find(needle, pos + 1)
        if pos == -1:
            break
        hits.append(pos)
    return hits

if __name__ == "__main__":
    print("Строка:", text)
```
